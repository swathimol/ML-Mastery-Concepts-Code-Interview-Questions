# Introduction to Classification
Classification is a supervised learning technique used to predict categorical outcomes. It is used when the targret variable is discrete. Examples include predicting whether an email is spam or not, whether a loan will be approved or not, etc.

### Types of Classification:

* Binary Classification: Only two possible classes (e.g., yes/no, spam/not spam).
* Multi-class Classification: More than two classes (e.g., types of fruits).

### Common Algorithms:



* Logistic Regression
* Decision Trees
* Random Forest
* Support Vector Machines (SVM)
* k-Nearest Neighbors (k-NN)
* Neural Networks

```python
import seaborn as sns
data=sns.load_dataset("iris")
```

```python
data
```

```python
data.info()
```

```python
# Define features and target
X = data.drop('species', axis=1)   #FEATURES, Q
X
```

```python
y = data['species']   #TARGET/ OUTPUT/ANS
y
```

```python
y.value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder

# Encode the target variable
le = LabelEncoder()
y = le.fit_transform(y)
y
```

```python
le.inverse_transform([2,0,1])
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
y_test.shape
```

```python
# Initialize and train the model
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
```

```python
# Make predictions
y_pred = logreg.predict(X_test)
y_pred
```

```python
y_test
```

```python
cm=confusion_matrix(y_test, y_pred)
import matplotlib.pyplot as plt
sns.heatmap(cm, annot=True,  xticklabels=le.classes_, yticklabels=le.classes_)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix for the Iris Dataset')
plt.show()
```

```python
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

```

```python
# Classification report
cr = classification_report(y_test, y_pred)
print("Classification Report:")
print(cr)
```

Precision is about correct positive predictions.
Recall is about finding all the positives.

```python

# Accuracy score
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy Score:")
print(accuracy)
```
